```python
dbutils.widgets.text("source_container", "")
dbutils.widgets.text("sink_container", "")
dbutils.widgets.text("csv_file", "")
dbutils.widgets.text("sink_file", "")
```

```python
source_container = dbutils.widgets.get("source_container")
sink_container = dbutils.widgets.get("sink_container")
csv_file = dbutils.widgets.get("csv_file")
sink_file = dbutils.widgets.get("sink_file")
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

```python
df = spark.read.format("csv").option("inferSchema", "True").option("header", "True").load(source_container + csv_file)
```

```python
df.write.format("delta")\
        .mode("overwrite")\
        .save(sink_container + sink_file)
```
